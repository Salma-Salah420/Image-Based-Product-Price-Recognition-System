**preprocess data..It includes three phase:
1.firstly:Clean text data**

```python
text_cols = ['name','keyword','Brand Name','Flavor','Item Form']

for col in text_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()
```

**Clean numerical features**

```python
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df = df.dropna(subset=['price'])
```

```python
df['price']
```

**ItemWeight cleaning**

```python
def convert_weight(x):
    x = str(x).lower()

    if 'kilogram' in x:
        return float(x.split()[0])

    elif 'gram' in x:
        return float(x.split()[0]) / 1000

    elif 'liter' in x:
        return float(x.split()[0])

    elif 'milliliter' in x:
        return float(x.split()[0]) / 1000

    else:
        return None

df['Item Weight'] = df['Item Weight'].apply(convert_weight)
```

**Clean the UnitCount Feature**

```python
df['Unit Count'] = pd.to_numeric(df['Unit Count'], errors='coerce')
```

```python
X = df.drop('price', axis=1)   # features
y = df['price']                # target
```
